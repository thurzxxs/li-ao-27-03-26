```python
# estrutura de decisões (if, elif e else)

# escrever um programa que compara 2 numeros

# entrada
a = int (input("digite um numero inteiro: "))
b= int (input("digite um numero inteiro: "))

# processamento / saida

if a == b:
  print (f"{a}é igual a {b}")

else:
   print (f"{a} é diferente de {b}")

```

```python
# escrever um programa que verifica se o número digitado é par ou impar

# entrada

numero = int (input("Digite um número: "))

# processamento / saída

if numero % 2 == 0:

  print (f" O número {numero} é par.")

else:
  print (f"O número {numero} é impar")
```
